```python
import pandas as pd

df =pd.read_csv(
    "../data/steam-200k.csv",
    header=None,
    names=["id_player","Game","buy?","playtime","unused"]
    )
df.describe(include=[float],percentiles=[0.2,0.6])
total_user = df["id_player"].nunique()
print("le nb total d'utilisateur est",total_user)
total_game = df["Game"].nunique()
print("le nb total de jeu est",total_game)
total_interaction = df["buy?"].count()
print("le nb total d'interaction est",total_interaction)
game_owed = df.groupby(by="id_player",as_index=False).count().sort_values(by="Game",ascending=False).iloc[:, :2]
print("le nb de jeu possédé par les utilisateurs sont:\n",game_owed.head(3).reset_index(drop=True).to_string(header=False))
jeux_plus_own = (
    df[df["buy?"] == "purchase"].groupby(by="Game")["id_player"]
    .size()
    .reset_index(name="nb_joueur")
    .sort_values(by="nb_joueur", ascending=False)
)
print("les jeux les plus possédés sont:\n",jeux_plus_own.head(3).reset_index(drop=True).to_string(header=False))
jeux_plus_jouer = (
    df[df["buy?"] == "play"].groupby(by="Game")["playtime"]
    .sum()
    .reset_index(name="total_playtime")
    .sort_values(by="total_playtime", ascending=False)
)
print("les jeux les plus joués sont:\n",jeux_plus_jouer.head(3).reset_index(drop=True).to_string(header=False))
```

```python
#mini challenge
# le temps de jeu moyen par jeu ;
# les 10 jeux avec le temps de jeu moyen le plus élevé ;
# le temps de jeu moyen par utilisateur ;
# le nombre moyen de jeux possédés par utilisateur ;
# la médiane du nombre de jeux possédés par utilisateur.
import pandas as pd

df = pd.read_csv(
    "../data/steam-200k.csv",
    header = None,
    names=["id_player","game","buy?","playtime","unused"]
)
tps_moyen_par_jeu = (
    df[df["buy?"] == "play"].groupby(by="game")["playtime"]
    .mean()
    .reset_index(name="mean_playtime")
    .sort_values(by="mean_playtime",ascending=False)
    .round(2)
)
print("the mean time for each game is\n",tps_moyen_par_jeu.reset_index(drop=True).head(n=10))
mean_tps_by_user = (
    df[df["buy?"] == "play"].groupby(by="id_player")["playtime"]
    .mean()
    .reset_index(name="mean_playtime")
    .sort_values(by="mean_playtime",ascending=False)
    .round(2)
)
print("the mean time for each user is\n",mean_tps_by_user.reset_index(drop=True).head(n=10))
mean_own_game_each_user =(
    df[df["buy?"] == "purchase"]["buy?"]
    .count()
)
nb_user = df["id_player"].nunique()
print(f"the mean of game own by each user is {mean_own_game_each_user/nb_user:.2f}")
median_own_game_each_user = (
    df[df["buy?"] == "purchase"].groupby(by="id_player")["buy?"]
    .count()
    .median()
)
print("the median of game own by each user is ", median_own_game_each_user)

```

```python
#show a table with for each game the nb of player the total playtime and the mean_playtime
import pandas as pd

df = pd.read_csv(
    "../data/steam-200k.csv",
    header=None,
    names=["id_player","game","buy?","playtime","unused"]
)

resultat = (
    df[df["buy?"] == "play"].groupby(by="game")["playtime"]
    .agg(["count","sum","mean"])
    .reset_index(names=["game","total_player","total_playtime","mean_playtime"])
    .sort_values(by="total_player",ascending=False)
    .round(2)
)
print(resultat.head(n=10))
```
